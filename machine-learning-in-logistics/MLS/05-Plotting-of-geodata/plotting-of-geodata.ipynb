In the first task, route plots for ship movements are to be created. The provided file `Voyage_Data_RTM_HAM_Ships` is to be used for this. This includes AIS data from trips from Rotterdam to Hamburg and thus the following columns:

- `ShipID`: In order to identify the ships, but to anonymize the MMSIs, an ID was assigned to each ship, which starts consecutively from `1`
- `Latitude`: Geographical latitude
- `Longitude`: Geographical longitude
- `SOG`: Speed over ground in knots
- `COG`: Course over ground in degrees (°)
- `Time`: Time
- `StartTime`: Time of the start of a trip, i.e., time of departure from the starting port
- `EndTime`: Time of the end of a trip, i.e., time of arrival at the destination port
- `TravelTime`: Travel time from the starting port to the destination port in minutes
- `RemainingTravelTime`: Remaining travel time from the position of the geographical longitude and latitude to the destination port in minutes
